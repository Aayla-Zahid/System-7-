## AI, Machine Learning and Deep Learning
- **Artificial Intelligence (AI):** making machines do tasks that need human-like intelligence. This is the widest field.
- **Machine Learning (ML):** a part of AI where the computer learns patterns from data instead of being given fixed rules. Example: predicting house prices from past sales.
- **Deep Learning (DL):** a part of ML that uses neural networks with many layers. It needs a lot of data. Example: image recognition.

Deep Learning is inside Machine Learning, and Machine Learning is inside AI.

## Types of learning
| Type | Meaning | Example |
|---|---|---|
| Supervised | Learns from data that has the correct answers (labels) | Spam detection, house-price prediction |
| Unsupervised | Finds patterns in data without labels | Grouping customers (clustering) |
| Reinforcement | An agent learns from rewards and penalties for its actions | A game-playing agent |

## Dataset terms
A small house table is used as the example below.

```python
import pandas as pd

df = pd.DataFrame({
    'area_sqft': [1500, 900, 2200],
    'bedrooms':  [3, 2, 4],
    'age':       [10, 25, 5],
    'price':     [210000, 120000, 330000],
})
df
```

```python
X = df[['area_sqft', 'bedrooms', 'age']]   # features (inputs)
y = df['price']                            # label / target (what we predict)

print('Samples (rows)   :', df.shape[0])
print('Features (cols)  :', list(X.columns))
print('Target           :', y.name)
print('First sample     :', X.iloc[0].tolist(), '->', y.iloc[0])
```

- **Dataset:** the whole table.
- **Sample:** one row (one house).
- **Features:** the input columns (area_sqft, bedrooms, age).
- **Label / target:** the column we want to predict (price).
- **Regression vs classification:** if the target is a number it is regression, if it is a category it is classification.
