## 1. Lists

A list is an **ordered, mutable** collection that can hold items of any type.

### 1.1 Creating lists

```python
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "two", 3.0, True, None]
empty = []

# From other iterables
from_range = list(range(5))
from_string = list("hello")

print(fruits)
print(mixed)
print(from_range)
print(from_string)
```

### 1.2 Indexing and slicing

```python
letters = ["a", "b", "c", "d", "e", "f"]

print(letters[0])       # first
print(letters[-1])      # last
print(letters[1:4])     # index 1 up to (not including) 4
print(letters[:3])      # first three
print(letters[3:])      # from index 3 to end
print(letters[::2])     # every second item
print(letters[::-1])    # reversed copy
```

### 1.3 Modifying lists

```python
nums = [10, 20, 30]

nums.append(40)            # add to the end
nums.insert(1, 15)         # insert at index
nums.extend([50, 60])      # add many items
print(nums)

nums[0] = 5                # change by index
print(nums)

nums.remove(15)            # remove first matching value
last = nums.pop()          # remove & return last item
second = nums.pop(1)       # remove & return item at index
print(nums, "| popped:", last, second)

del nums[0]                # delete by index
print(nums)

nums.clear()               # empty the list
print(nums)
```

### 1.4 Useful list methods

```python
data = [5, 2, 9, 1, 5, 6]

print("count of 5:", data.count(5))
print("index of 9:", data.index(9))

data.sort()                       # in place
print("sorted:", data)

data.sort(reverse=True)
print("descending:", data)

data.reverse()
print("reversed:", data)

# sorted() returns a NEW list, leaving the original alone
original = [3, 1, 2]
new = sorted(original)
print(original, new)

# Built-in helpers
print(len(data), min(data), max(data), sum(data))
```

### 1.5 Copying — a common gotcha

```python
a = [1, 2, 3]
b = a            # same list, two names!
b.append(4)
print("a:", a, "| b:", b)

c = a.copy()     # independent (shallow) copy
c.append(99)
print("a:", a, "| c:", c)

# Shallow copy caveat with nested lists
import copy
nested = [[1, 2], [3, 4]]
shallow = nested.copy()
deep = copy.deepcopy(nested)
nested[0].append("X")
print("shallow:", shallow)   # affected
print("deep:   ", deep)      # not affected
```

### 1.6 List comprehensions

```python
squares = [x**2 for x in range(1, 6)]
evens = [x for x in range(10) if x % 2 == 0]
labels = ["even" if x % 2 == 0 else "odd" for x in range(5)]
pairs = [(x, y) for x in range(2) for y in range(2)]

print(squares)
print(evens)
print(labels)
print(pairs)
```

### 1.7 Looping, membership and nesting

```python
colors = ["red", "green", "blue"]

for i, color in enumerate(colors, start=1):
    print(i, color)

print("green" in colors)
print("pink" not in colors)

# zip pairs items from several lists
names = ["Ana", "Ben", "Cy"]
scores = [90, 85, 77]
for n, s in zip(names, scores):
    print(f"{n}: {s}")

# 2D list (matrix)
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
print(matrix[1][2])
print([row[0] for row in matrix])   # first column
```

## 2. Tuples

A tuple is an **ordered, immutable** collection. Once created, it can't be changed.

### 2.1 Creating tuples

```python
point = (3, 4)
single = (42,)          # trailing comma needed for a 1-item tuple!
not_a_tuple = (42)      # just an int
no_parens = 1, 2, 3     # parentheses are optional
empty = ()

print(type(single), type(not_a_tuple), type(no_parens))
print(tuple([1, 2, 3]))
```

### 2.2 Accessing and immutability

```python
t = ("a", "b", "c", "d")
print(t[0], t[-1], t[1:3])
print(len(t), t.count("a"), t.index("c"))

try:
    t[0] = "z"
except TypeError as e:
    print("Error:", e)

# Workaround: convert, change, convert back
tmp = list(t)
tmp[0] = "z"
t = tuple(tmp)
print(t)
```

### 2.3 Packing and unpacking

```python
person = ("Alice", 30, "Lahore")
name, age, city = person
print(name, age, city)

# Extended unpacking with *
first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)

# Swap values without a temp variable
x, y = 1, 2
x, y = y, x
print(x, y)

# Functions can return multiple values (as a tuple)
def min_max(values):
    return min(values), max(values)

lo, hi = min_max([4, 9, 2, 7])
print(lo, hi)
```

### 2.4 Why use tuples?

```python
# 1) Hashable -> can be dictionary keys or set members (lists cannot)
locations = {(31.52, 74.35): "Lahore", (24.86, 67.01): "Karachi"}
print(locations[(31.52, 74.35)])

# 2) Slightly smaller and faster than lists
import sys
print("list :", sys.getsizeof([1, 2, 3]), "bytes")
print("tuple:", sys.getsizeof((1, 2, 3)), "bytes")

# 3) Signals "this data shouldn't change"
# Note: a tuple holding a mutable object can still change that object
box = ([1, 2], "x")
box[0].append(3)
print(box)

# Named tuples give fields readable names
from collections import namedtuple
Point = namedtuple("Point", ["x", "y"])
p = Point(2, 5)
print(p.x, p.y, p)
```

## 3. Dictionaries

A dictionary stores **key → value** pairs. Keys must be unique and hashable. Since Python 3.7, dictionaries keep **insertion order**.

### 3.1 Creating dictionaries

```python
student = {"name": "Sara", "age": 20, "major": "CS"}
via_dict = dict(name="Sara", age=20)
from_pairs = dict([("a", 1), ("b", 2)])
from_keys = dict.fromkeys(["x", "y", "z"], 0)
empty = {}

print(student)
print(via_dict)
print(from_pairs)
print(from_keys)
```

### 3.2 Accessing values

```python
print(student["name"])
print(student.get("age"))
print(student.get("grade"))            # None instead of an error
print(student.get("grade", "N/A"))     # custom default

try:
    student["grade"]
except KeyError as e:
    print("KeyError:", e)
```

### 3.3 Adding, updating and removing

```python
d = {"a": 1, "b": 2}

d["c"] = 3                    # add
d["a"] = 100                  # update
d.update({"d": 4, "b": 20})   # bulk add/update
print(d)

print(d.pop("c"))             # remove & return
print(d.pop("zzz", None))     # safe pop with default
del d["d"]
print(d)

d.setdefault("e", 5)          # add only if missing
print(d)

key, value = d.popitem()      # removes the last inserted pair
print(key, value)
```

### 3.4 Keys, values, items and looping

```python
scores = {"Ana": 90, "Ben": 85, "Cy": 77}

print(list(scores.keys()))
print(list(scores.values()))
print(list(scores.items()))

for name, score in scores.items():
    print(f"{name} -> {score}")

print("Ana" in scores)        # checks keys
print(90 in scores.values())  # checks values
```

### 3.5 Dictionary comprehensions and merging

```python
squares = {n: n**2 for n in range(1, 6)}
print(squares)

# Filter
high = {k: v for k, v in scores.items() if v >= 80}
print(high)

# Swap keys and values
print({v: k for k, v in scores.items()})

# Merge (Python 3.9+)
a = {"x": 1, "y": 2}
b = {"y": 99, "z": 3}
print(a | b)              # b wins on conflicts
print({**a, **b})         # older syntax, same result
```

### 3.6 Nested dictionaries

```python
company = {
    "alice": {"role": "dev", "skills": ["python", "sql"]},
    "bob":   {"role": "designer", "skills": ["figma"]},
}

print(company["alice"]["role"])
print(company["alice"]["skills"][0])

company["bob"]["skills"].append("css")
print(company["bob"])

for person, info in company.items():
    print(person, "->", info["role"], len(info["skills"]), "skills")
```

### 3.7 Practical patterns: counting and grouping

```python
# Count word frequency
text = "the cat and the hat and the bat"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)

# Same idea using Counter
from collections import Counter
print(Counter(text.split()).most_common(2))

# Group items with defaultdict
from collections import defaultdict
words = ["apple", "avocado", "banana", "blueberry", "cherry"]
by_letter = defaultdict(list)
for w in words:
    by_letter[w[0]].append(w)
print(dict(by_letter))
```

## 4. Sets

A set is an **unordered collection of unique, hashable items**. Great for removing duplicates and fast membership tests.

### 4.1 Creating sets

```python
colors = {"red", "green", "blue"}
from_list = set([1, 2, 2, 3, 3, 3])
from_string = set("hello")
empty = set()            # {} would create an empty dict!

print(colors)
print(from_list)
print(from_string)
print(type({}), type(empty))
```

### 4.2 Adding and removing

```python
s = {1, 2, 3}

s.add(4)
s.add(2)                 # duplicate, ignored
s.update([5, 6, 7])
print(s)

s.remove(7)              # KeyError if missing
s.discard(100)           # no error if missing
x = s.pop()              # removes an arbitrary item
print(s, "| popped:", x)

try:
    s.remove(999)
except KeyError as e:
    print("KeyError:", e)
```

### 4.3 Set operations

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7}

print("union:               ", a | b, a.union(b))
print("intersection:        ", a & b, a.intersection(b))
print("difference (a - b):  ", a - b, a.difference(b))
print("symmetric difference:", a ^ b, a.symmetric_difference(b))
```

### 4.4 Subset and superset checks

```python
small = {1, 2}
big = {1, 2, 3, 4}

print(small <= big, small.issubset(big))
print(big >= small, big.issuperset(small))
print({1, 2}.isdisjoint({3, 4}))   # no elements in common
```

### 4.5 Fast membership and de-duplication

```python
import time

n = 200_000
as_list = list(range(n))
as_set = set(as_list)
target = n - 1

t0 = time.perf_counter()
target in as_list
t_list = time.perf_counter() - t0

t0 = time.perf_counter()
target in as_set
t_set = time.perf_counter() - t0

print(f"list lookup: {t_list:.6f}s")
print(f"set lookup : {t_set:.6f}s")

# Remove duplicates (order not guaranteed)
dupes = [3, 1, 2, 3, 2, 1]
print(list(set(dupes)))

# Remove duplicates but keep order
print(list(dict.fromkeys(dupes)))
```

### 4.6 Set comprehensions and frozenset

```python
print({x % 3 for x in range(10)})

# frozenset is an immutable set (hashable)
fs = frozenset([1, 2, 3])
lookup = {fs: "immutable set as a key"}
print(lookup[frozenset([3, 2, 1])])

# Real example: find common interests
alice = {"python", "music", "chess"}
bob = {"music", "football", "python"}
print("shared:", alice & bob)
print("only Alice:", alice - bob)
```

## 5. Choosing the Right Structure

| If you need… | Use |
|---|---|
| An ordered, changeable sequence | **list** |
| Fixed data, a dictionary key, or multiple return values | **tuple** |
| Fast lookup by a label/key | **dict** |
| Unique items, fast membership tests, or set math | **set** |


### Key takeaways
- **Lists** are your default flexible sequence; watch out for aliasing (`b = a`) vs copying.
- **Tuples** are immutable, which makes them hashable and safe for fixed data.
- **Dictionaries** give fast key-based lookup; prefer `.get()` and `defaultdict` for safe access.
- **Sets** enforce uniqueness and make intersection/union/difference one-liners.

## 6. Mini Challenge

Combine everything: given the data below, produce a dictionary mapping each **category** to a **sorted list of unique items**.

```python
purchases = [
    ("fruit", "apple"), ("veg", "carrot"), ("fruit", "banana"),
    ("fruit", "apple"), ("veg", "spinach"), ("veg", "carrot"),
]

# Your solution here
# Expected: {'fruit': ['apple', 'banana'], 'veg': ['carrot', 'spinach']}
```

```python
# One possible solution
from collections import defaultdict

grouped = defaultdict(set)
for category, item in purchases:
    grouped[category].add(item)

result = {cat: sorted(items) for cat, items in grouped.items()}
print(result)
```
