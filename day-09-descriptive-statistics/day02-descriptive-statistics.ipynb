Topics: mean, median, mode, variance and standard deviation.

Data: marks of 10 students.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

scores = [72, 85, 90, 85, 60, 78, 85, 92, 66, 78]
s = pd.Series(scores)

print('Mean            :', np.mean(scores))
print('Median          :', np.median(scores))
print('Mode            :', s.mode().tolist())
print('Variance (pop)  :', round(np.var(scores), 2))
print('Std dev  (pop)  :', round(np.std(scores), 3))
print('Variance (samp) :', round(np.var(scores, ddof=1), 3))
print('Std dev  (samp) :', round(np.std(scores, ddof=1), 3))
```

## Calculation by hand
- Mean = 791 / 10 = 79.1
- Median: sorted data is 60, 66, 72, 78, 78, 85, 85, 85, 90, 92. The middle two values are 78 and 85, so median = (78 + 85) / 2 = 81.5
- Mode: 85, because it appears 3 times.
- Variance = average of (value - mean) squared.
- Standard deviation = square root of variance.

```python
mean = sum(scores) / len(scores)
manual_var = sum((x - mean) ** 2 for x in scores) / len(scores)
print('Manual variance      :', round(manual_var, 2))
print('Manual std deviation :', round(manual_var ** 0.5, 3))
```

```python
plt.hist(scores, bins=6, edgecolor='black')
plt.axvline(np.mean(scores), color='red', linestyle='--', label='mean')
plt.axvline(np.median(scores), color='green', linestyle=':', label='median')
plt.xlabel('Marks')
plt.ylabel('Number of students')
plt.title('Distribution of marks')
plt.legend()
plt.show()
```

## Mean vs median with an extreme value
The mean changes a lot with an extreme value, the median does not.

```python
with_outlier = scores + [10]
print('Mean   without / with 10:', np.mean(scores), '/', round(np.mean(with_outlier), 2))
print('Median without / with 10:', np.median(scores), '/', np.median(with_outlier))
```

## Notes
- Population variance divides by n, sample variance divides by n - 1 (`ddof=1`). NumPy uses the first by default and Pandas uses the second.
