### 1.2 Comments

### 1.3 Indentation

Indentation tells Python which lines belong together. The standard is **4 spaces** per level.

````python
age = 20

if age >= 18:
    print("You are an adult")      # indented: runs only if the condition is True
    print("You can vote")          # same block
print("This line always runs")     # not indented: outside the if block
````

Wrong indentation causes an `IndentationError`:

```python
if age >= 18:
print("Oops")   # IndentationError: expected an indented block
```

### 1.4 Multiple statements and line continuation

````python
# Two statements on one line need a semicolon (not recommended style)
a = 1; b = 2

# A long statement can be split across lines using a backslash
total = 1 + 2 + 3 + \
        4 + 5 + 6
print(total)

# Inside brackets (), [], {} you can break lines freely
numbers = [
    10, 20, 30,
    40, 50,
]
print(numbers)
````

### 1.5 Naming rules and conventions

Rules (must follow):
- Can contain letters, digits and underscores
- Cannot start with a digit
- Cannot contain spaces or special characters like `-`, `$`, `@`
- Cannot be a Python keyword (`if`, `for`, `class`, ...)

Convention (PEP 8): use `snake_case` for variable and function names, e.g. `student_name`.

## 2. Variables

A variable is a name that refers to a value stored in memory. You create one by assigning a value with `=`. There is no need to declare a type first, because Python is **dynamically typed**.

### 2.1 Creating variables

````python
name = "Ayesha"
age = 21
height = 5.6
is_student = True

print(name)
print(age)
print(height)
print(is_student)
````

### 2.2 Changing a variable's value (and type)

````python
x = 10
print(x, type(x))

x = "now I am text"
print(x, type(x))
````

### 2.3 Multiple assignment

````python
# Different values to different variables
a, b, c = 1, 2, 3
print(a, b, c)

# Same value to several variables
x = y = z = 0
print(x, y, z)

# Swapping two variables
a, b = b, a
print(a, b)
````

### 2.4 Constants (by convention)

````python
# Python has no true constants. By convention, ALL_CAPS means "don't change this".
PI = 3.14159
MAX_USERS = 100
print(PI, MAX_USERS)
````

### 2.5 Useful functions: `type()`, `id()`, `del`

````python
score = 95
print(type(score))   # what type is it?
print(id(score))     # memory identity of the object

del score            # delete the variable
# print(score)       # would raise NameError: name 'score' is not defined
````

## 3. Data Types

Every value in Python has a type. The built-in types you will use most:

| Category | Types |
|---|---|
| Numeric | `int`, `float`, `complex` |
| Text | `str` |
| Boolean | `bool` |
| Sequence | `list`, `tuple`, `range` |
| Mapping | `dict` |
| Set | `set`, `frozenset` |
| None | `NoneType` |

### 3.1 Numbers: `int`, `float`, `complex`

````python
whole = 42              # int: whole numbers of any size
decimal = 3.14          # float: numbers with a decimal point
complex_num = 2 + 3j    # complex: real + imaginary part

print(whole, type(whole))
print(decimal, type(decimal))
print(complex_num, type(complex_num))

big = 12345678901234567890 ** 2   # ints have no size limit
print(big)
````

### 3.2 Strings: `str`

````python
s1 = 'single quotes'
s2 = "double quotes"
s3 = """triple quotes
can span
multiple lines"""

print(s1)
print(s2)
print(s3)
````

**Indexing and slicing** (indexes start at 0; negative indexes count from the end)

````python
word = "Python"

print(word[0])      # P
print(word[-1])     # n
print(word[0:3])    # Pyt  (start included, end excluded)
print(word[2:])     # thon
print(word[::-1])   # nohtyP (reversed)
print(len(word))    # 6
````

**Common string methods** (strings are immutable, so methods return a new string)

````python
text = "  Hello, Python World!  "

print(text.upper())
print(text.lower())
print(text.strip())                 # remove leading/trailing spaces
print(text.replace("World", "All"))
print(text.split(","))              # split into a list
print("Python" in text)             # membership test
print(text.strip().startswith("Hello"))
print(text.count("o"))
````

**String formatting**

````python
name = "Ali"
age = 22

# f-strings (recommended)
print(f"My name is {name} and I am {age} years old.")

# format() method
print("My name is {} and I am {} years old.".format(name, age))

# Formatting numbers
pi = 3.14159265
print(f"Pi to 2 decimals: {pi:.2f}")
print(f"Padded number: {42:05d}")
print(f"With commas: {1234567:,}")
````

### 3.3 Boolean: `bool`

````python
is_raining = True
is_sunny = False
print(is_raining, type(is_raining))

# Comparisons produce booleans
print(5 > 3)
print(5 == 3)

# Truthy and falsy values
print(bool(0), bool(""), bool([]), bool(None))      # all False
print(bool(1), bool("hi"), bool([0]))               # all True
````

### 3.4 List: ordered and **mutable**

````python
fruits = ["apple", "banana", "cherry"]

print(fruits)
print(fruits[0])
fruits.append("mango")        # add to end
fruits.insert(1, "orange")    # insert at index
fruits.remove("banana")       # remove by value
last = fruits.pop()           # remove and return last item
fruits[0] = "grape"           # change an item
print(fruits, last)
print(len(fruits))

nums = [5, 2, 9, 1]
nums.sort()
print(nums)
````

### 3.5 Tuple: ordered and **immutable**

````python
point = (10, 20)
print(point, point[0])

# point[0] = 99   # TypeError: tuples cannot be changed

x, y = point      # unpacking
print(x, y)

single = (5,)     # a one-item tuple needs a trailing comma
print(type(single))
````

### 3.6 Dictionary: key-value pairs

````python
student = {
    "name": "Sara",
    "age": 20,
    "course": "Data Science",
}

print(student["name"])
print(student.get("grade", "N/A"))   # safe lookup with default
student["age"] = 21                  # update
student["city"] = "Lahore"           # add
del student["course"]                # remove

print(student)
print(student.keys())
print(student.values())
print(student.items())
````

### 3.7 Set: unordered, **unique** items

````python
numbers = {1, 2, 3, 3, 2, 1}
print(numbers)          # duplicates removed

numbers.add(4)
numbers.discard(1)
print(numbers)

a = {1, 2, 3}
b = {3, 4, 5}
print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
````

### 3.8 `range` and `None`

````python
print(list(range(5)))          # 0..4
print(list(range(2, 10, 2)))   # 2, 4, 6, 8

result = None                  # represents "no value"
print(result, type(result))
````

### 3.9 Type conversion (casting)

````python
print(int("25") + 5)        # str -> int
print(float("3.5") * 2)     # str -> float
print(str(100) + " marks")  # int -> str
print(int(7.9))             # float -> int (truncates, does not round)
print(list("abc"))          # str -> list
print(tuple([1, 2, 3]))     # list -> tuple
print(set([1, 1, 2]))       # list -> set

# Invalid conversions raise ValueError
# int("hello")
````

### 3.10 Mutable vs immutable

| Immutable (cannot change in place) | Mutable (can change in place) |
|---|---|
| `int`, `float`, `bool`, `str`, `tuple`, `frozenset` | `list`, `dict`, `set` |

## 4. Operators

Operators perform operations on values (operands).

### 4.1 Arithmetic operators

````python
a, b = 17, 5

print("a + b  =", a + b)    # addition
print("a - b  =", a - b)    # subtraction
print("a * b  =", a * b)    # multiplication
print("a / b  =", a / b)    # division (always float)
print("a // b =", a // b)   # floor division
print("a % b  =", a % b)    # modulus (remainder)
print("a ** b =", a ** b)   # exponent (power)
````

### 4.2 Comparison operators (return `True` / `False`)

````python
x, y = 10, 20

print(x == y)   # equal to
print(x != y)   # not equal to
print(x > y)    # greater than
print(x < y)    # less than
print(x >= 10)  # greater than or equal to
print(y <= 15)  # less than or equal to

print(1 < x < 15)   # chained comparison
````

### 4.3 Assignment operators

````python
n = 10
n += 5    # n = n + 5
print(n)
n -= 3    # n = n - 3
print(n)
n *= 2    # n = n * 2
print(n)
n /= 4    # n = n / 4
print(n)
n //= 2
print(n)
n %= 2
print(n)
n = 3
n **= 3
print(n)
````

### 4.4 Logical operators: `and`, `or`, `not`

````python
age = 25
has_id = True

print(age >= 18 and has_id)     # True only if both are True
print(age < 18 or has_id)       # True if at least one is True
print(not has_id)               # reverses the value

# Truth table
for p in (True, False):
    for q in (True, False):
        print(f"{p!s:5} and {q!s:5} = {p and q!s:5} | {p!s:5} or {q!s:5} = {p or q}")
````

### 4.5 Identity operators: `is`, `is not`

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)   # True: same values
print(a is b)   # False: different objects in memory
print(a is c)   # True: same object
print(a is not b)

# Common use: checking for None
value = None
print(value is None)
````

### 4.6 Membership operators: `in`, `not in`

````python
print("a" in "banana")
print(3 in [1, 2, 3])
print("x" not in "banana")
print("name" in {"name": "Ali", "age": 20})   # checks keys
````

### 4.7 Bitwise operators

````python
a = 12   # 1100 in binary
b = 10   # 1010 in binary

print(a & b)    # AND  -> 8  (1000)
print(a | b)    # OR   -> 14 (1110)
print(a ^ b)    # XOR  -> 6  (0110)
print(~a)       # NOT  -> -13
print(a << 1)   # left shift  -> 24
print(a >> 1)   # right shift -> 6
print(bin(a), bin(b))
````

### 4.8 Operator precedence

From highest to lowest (simplified):

1. `()` parentheses
2. `**`
3. `+x`, `-x`, `~x`
4. `*`, `/`, `//`, `%`
5. `+`, `-`
6. Comparisons, `is`, `in`
7. `not`
8. `and`
9. `or`

````python
print(2 + 3 * 4)        # 14, multiplication first
print((2 + 3) * 4)      # 20, parentheses first
print(2 ** 3 ** 2)      # 512, ** is right-associative: 2 ** (3 ** 2)
print(not True or True) # True: not is applied before or
````

## 5. Input / Output

### 5.1 Output with `print()`

````python
print("Hello")
print("Hello", "World")               # multiple values, separated by a space
print("a", "b", "c", sep="-")         # custom separator
print("Loading", end="...")           # custom ending (no newline)
print("done")
print("Line 1\nLine 2")               # \n = new line
print("Tab\tseparated")               # \t = tab
print("She said \"Hi\"")              # escaped quote
print(r"Raw string: C:\new\folder")  # r"" ignores escapes
````

**Formatted output**

````python
name = "Zain"
marks = 87.456

print(f"Name: {name}, Marks: {marks:.1f}")
print("Name: %s, Marks: %.2f" % (name, marks))   # old style
print("Name: {}, Marks: {}".format(name, marks))
print(f"{'Left':<10}|{'Center':^10}|{'Right':>10}")  # alignment
````

### 5.2 Input with `input()`

`input()` pauses the program, waits for the user to type something and press Enter, and **always returns a string**. Convert it with `int()`, `float()` etc. if you need a number.

````python
name = input("Enter your name: ")
print("Hello,", name)
````

````python
age = int(input("Enter your age: "))
print("Next year you will be", age + 1)
````

````python
# Reading several values on one line, e.g. "10 20"
a, b = map(int, input("Enter two numbers separated by space: ").split())
print("Sum =", a + b)
````

### 5.3 Handling invalid input

````python
try:
    number = int(input("Enter a whole number: "))
    print("You entered", number)
except ValueError:
    print("That was not a valid whole number.")
````

### 5.4 File input / output (introduction)

Use `open()` with a mode: `"w"` write, `"r"` read, `"a"` append. The `with` statement closes the file automatically.

````python
# Write to a file
with open("sample.txt", "w") as f:
    f.write("Hello from Python!\n")
    f.write("This is line 2.\n")

# Read the whole file
with open("sample.txt", "r") as f:
    content = f.read()
print(content)

# Read line by line
with open("sample.txt", "r") as f:
    for line in f:
        print(line.strip())

# Append
with open("sample.txt", "a") as f:
    f.write("This is line 3.\n")
````

## 6. Putting It Together: Mini Programs

### 6.1 Simple calculator

````python
num1 = float(input("First number: "))
num2 = float(input("Second number: "))

print(f"{num1} + {num2} = {num1 + num2}")
print(f"{num1} - {num2} = {num1 - num2}")
print(f"{num1} * {num2} = {num1 * num2}")
if num2 != 0:
    print(f"{num1} / {num2} = {num1 / num2:.2f}")
else:
    print("Cannot divide by zero")
````

### 6.2 Student report

````python
name = input("Student name: ")
m1 = float(input("Marks in subject 1: "))
m2 = float(input("Marks in subject 2: "))
m3 = float(input("Marks in subject 3: "))

total = m1 + m2 + m3
average = total / 3
passed = average >= 50

print("\n--- Report ---")
print(f"Name    : {name}")
print(f"Total   : {total}")
print(f"Average : {average:.2f}")
print(f"Passed  : {passed}")
````

### 6.3 Temperature converter

````python
celsius = float(input("Temperature in Celsius: "))
fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius}°C = {fahrenheit:.1f}°F")
````
