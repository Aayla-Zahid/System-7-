## 1. Conditions

Conditions let a program make decisions. Python runs a block of code only when a condition is `True`.
Remember: blocks are defined by **indentation** (4 spaces).

### 1.1 `if` statement

````python
temperature = 35

if temperature > 30:
    print("It's a hot day")
    print("Drink plenty of water")

print("This line runs no matter what")
````

### 1.2 `if ... else`

````python
age = 16

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote yet")
````

### 1.3 `if ... elif ... else`

Python checks conditions from top to bottom and runs **only the first** one that is `True`.

````python
marks = 78

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "F"

print(f"Marks: {marks}, Grade: {grade}")
````

### 1.4 Comparison and logical operators in conditions

````python
age = 25
has_license = True

if age >= 18 and has_license:
    print("You can drive")

day = "Sunday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend")

is_raining = False
if not is_raining:
    print("No umbrella needed")

# Chained comparison
score = 65
if 50 <= score < 75:
    print("Average score")
````

### 1.5 Membership and identity checks

````python
vowels = "aeiou"
letter = "e"

if letter in vowels:
    print(f"{letter} is a vowel")

value = None
if value is None:
    print("No value provided")

fruits = ["apple", "banana"]
if "mango" not in fruits:
    print("Mango is not in the list")
````

### 1.6 Truthy and falsy values

Python treats these as `False`: `0`, `0.0`, `""` (empty string), `[]`, `()`, `{}`, `set()`, `None`.
Almost everything else is `True`.

````python
name = ""
if name:
    print("Hello,", name)
else:
    print("Name is empty")

items = [1, 2, 3]
if items:
    print("The list has", len(items), "items")
````

### 1.7 Nested `if`

````python
age = 22
has_ticket = True

if age >= 18:
    if has_ticket:
        print("Welcome to the event")
    else:
        print("Please buy a ticket")
else:
    print("Adults only")
````

### 1.8 Conditional expression (ternary operator)

A one-line `if ... else` that produces a value.

````python
age = 20
status = "adult" if age >= 18 else "minor"
print(status)

n = 7
print("even" if n % 2 == 0 else "odd")
````

### 1.9 `match` statement (Python 3.10+)

A cleaner way to compare one value against many patterns.

````python
import sys
print(sys.version)

if sys.version_info >= (3, 10):
    command = "stop"

    match command:
        case "start":
            print("Starting...")
        case "stop":
            print("Stopping...")
        case "pause" | "wait":
            print("Pausing...")
        case _:
            print("Unknown command")
else:
    print("match needs Python 3.10 or newer")
````

### 1.10 Example: user input with conditions

````python
number = int(input("Enter a number: "))

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
````

## 2. Loops

A loop repeats a block of code. Python has two kinds:

- `for` loop: repeat for each item in a sequence (or a fixed number of times)
- `while` loop: repeat as long as a condition is `True`

### 2.1 `for` loop over a list

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

### 2.2 `for` loop over a string, tuple, dict and set

````python
# String: one character at a time
for ch in "Python":
    print(ch, end=" ")
print()

# Tuple
for n in (10, 20, 30):
    print(n)

# Dictionary
student = {"name": "Sara", "age": 20, "city": "Lahore"}

for key in student:                      # keys
    print(key)

for value in student.values():           # values
    print(value)

for key, value in student.items():       # both
    print(f"{key} -> {value}")
````

### 2.3 Using `enumerate()` to get index and value

````python
colors = ["red", "green", "blue"]

for index, color in enumerate(colors):
    print(index, color)

# Start counting from 1
for index, color in enumerate(colors, start=1):
    print(f"{index}. {color}")
````

### 2.4 Looping over two lists with `zip()`

````python
names = ["Ali", "Sara", "Zain"]
marks = [85, 92, 78]

for name, mark in zip(names, marks):
    print(f"{name} scored {mark}")
````

### 2.5 `while` loop

````python
count = 1

while count <= 5:
    print("Count is", count)
    count += 1          # without this the loop would never end!

print("Loop finished")
````

### 2.6 `while` loop with user input

````python
password = ""

while password != "python123":
    password = input("Enter password: ")

print("Access granted")
````

### 2.7 `break`: exit the loop early

````python
for number in range(1, 20):
    if number == 6:
        print("Found 6, stopping")
        break
    print(number)
````

### 2.8 `continue`: skip to the next iteration

````python
for number in range(1, 11):
    if number % 2 == 0:
        continue            # skip even numbers
    print(number)
````

### 2.9 `pass`: a placeholder that does nothing

````python
for number in range(5):
    if number == 3:
        pass                # TODO: add code later
    print(number)
````

### 2.10 `else` with loops

The `else` block of a loop runs only if the loop finished **without** hitting `break`.

````python
numbers = [3, 5, 7, 9]

for n in numbers:
    if n % 2 == 0:
        print("Found an even number:", n)
        break
else:
    print("No even numbers found")
````

### 2.11 Nested loops

````python
# Multiplication table (1 to 3)
for i in range(1, 4):
    for j in range(1, 6):
        print(f"{i} x {j} = {i * j}")
    print("---")
````

### 2.12 Printing patterns

````python
# Right-angled triangle
for i in range(1, 6):
    print("*" * i)

print()

# Pyramid
rows = 5
for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
````

### 2.13 Common loop patterns

````python
numbers = [4, 8, 15, 16, 23, 42]

# Sum
total = 0
for n in numbers:
    total += n
print("Sum:", total)

# Largest value
largest = numbers[0]
for n in numbers:
    if n > largest:
        largest = n
print("Largest:", largest)

# Count items matching a condition
count = 0
for n in numbers:
    if n % 2 == 0:
        count += 1
print("Even numbers:", count)

# Build a new list
squares = []
for n in numbers:
    squares.append(n ** 2)
print("Squares:", squares)
````

### 2.14 List comprehension (a shorter way to write loops)

````python
numbers = [1, 2, 3, 4, 5, 6]

squares = [n ** 2 for n in numbers]
print(squares)

evens = [n for n in numbers if n % 2 == 0]
print(evens)

labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print(labels)
````

### 2.15 Avoiding infinite loops

A `while` loop whose condition never becomes `False` runs forever. Always make sure something inside the loop changes the condition, or use `break`.

```python
# Infinite loop, don't run this!
while True:
    print("forever")
```

`while True` is fine when you use `break` to exit:

````python
attempts = 0

while True:
    attempts += 1
    print("Attempt", attempts)
    if attempts == 3:
        break
````

## 3. `range()`

`range()` produces a sequence of integers. It is most often used with `for` loops.

| Form | Meaning |
|---|---|
| `range(stop)` | from 0 up to, but **not including**, `stop` |
| `range(start, stop)` | from `start` up to, but not including, `stop` |
| `range(start, stop, step)` | same, moving by `step` each time |

### 3.1 `range(stop)`

````python
for i in range(5):
    print(i)

print(list(range(5)))
````

### 3.2 `range(start, stop)`

````python
for i in range(1, 6):
    print(i)

print(list(range(10, 15)))
````

### 3.3 `range(start, stop, step)`

````python
print(list(range(0, 20, 5)))    # step of 5
print(list(range(2, 11, 2)))    # even numbers
print(list(range(1, 10, 2)))    # odd numbers
````

### 3.4 Counting backwards (negative step)

````python
print(list(range(10, 0, -1)))   # 10 down to 1
print(list(range(10, 0, -3)))   # 10, 7, 4, 1

for i in range(5, 0, -1):
    print(i)
print("Liftoff!")
````

### 3.5 Useful facts about `range`

````python
r = range(1, 10, 2)

print(r)                # a range object, not a list
print(list(r))          # convert to a list to see the values
print(len(r))           # number of items
print(r[0], r[-1])      # indexing works
print(5 in r)           # membership test
print(sum(range(1, 101)))   # sum of 1 to 100

# range is memory efficient: it doesn't store every number
big = range(1_000_000_000)
print(len(big))
````

### 3.6 `range()` with `len()` to loop over indexes

````python
names = ["Ali", "Sara", "Zain"]

for i in range(len(names)):
    print(i, names[i])
````

### 3.7 Examples using `range()`

````python
# Multiplication table of a number
n = 7
for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
````

````python
# Factorial of a number
n = 5
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print(f"{n}! = {factorial}")
````

````python
# Fibonacci series (first 10 terms)
a, b = 0, 1
for _ in range(10):
    print(a, end=" ")
    a, b = b, a + b
print()
````

````python
# Prime numbers between 2 and 30
for num in range(2, 31):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=" ")
print()
````

## 4. Functions

A function is a named, reusable block of code. You **define** it once with `def` and **call** it whenever needed.

Benefits: less repetition, easier to read, easier to test.

### 4.1 Defining and calling a function

````python
def greet():
    print("Hello! Welcome to Python.")

greet()
greet()
````

### 4.2 Parameters and arguments

````python
def greet(name):                 # name is a parameter
    print(f"Hello, {name}!")

greet("Ali")                     # "Ali" is an argument
greet("Sara")
````

````python
def add(a, b):
    print(a + b)

add(5, 3)
add(10, 20)
````

### 4.3 The `return` statement

`return` sends a value back to the caller and ends the function.

````python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
print(add(10, 20) * 2)
````

**`print` vs `return`**: `print` only shows text, `return` gives you a value you can use later.

````python
def square_print(n):
    print(n * n)

def square_return(n):
    return n * n

x = square_print(4)
y = square_return(4)

print("x =", x)    # None, because the function returned nothing
print("y =", y)    # 16
````

### 4.4 Returning multiple values

````python
def min_max(numbers):
    return min(numbers), max(numbers)    # returns a tuple

smallest, largest = min_max([4, 9, 1, 7])
print(smallest, largest)
````

### 4.5 Default parameter values

````python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Ali")                 # uses default greeting
greet("Ali", "Good morning") # overrides the default
````

### 4.6 Keyword arguments

Pass arguments by name, so their order does not matter.

````python
def describe_person(name, age, city):
    print(f"{name} is {age} years old and lives in {city}.")

describe_person("Sara", 21, "Lahore")                  # positional
describe_person(city="Karachi", name="Zain", age=22)   # keyword
````

### 4.7 `*args`: any number of positional arguments

````python
def total(*numbers):
    print(numbers)          # collected into a tuple
    return sum(numbers)

print(total(1, 2))
print(total(1, 2, 3, 4, 5))
````

### 4.8 `**kwargs`: any number of keyword arguments

````python
def show_profile(**details):
    print(details)          # collected into a dictionary
    for key, value in details.items():
        print(f"{key}: {value}")

show_profile(name="Ali", age=20, city="Lahore")
````

### 4.9 Docstrings

A string right after the `def` line that describes the function. View it with `help()`.

````python
def area_of_circle(radius):
    """Return the area of a circle with the given radius."""
    return 3.14159 * radius ** 2

print(area_of_circle(5))
help(area_of_circle)
````

### 4.10 Variable scope: local vs global

````python
message = "I am global"

def show():
    message = "I am local"      # a new local variable
    print(message)

show()
print(message)                  # the global one is unchanged
````

````python
counter = 0

def increment():
    global counter              # tell Python to use the global variable
    counter += 1

increment()
increment()
print(counter)
````

Variables created inside a function exist only while it runs:

```python
def f():
    x = 10

f()
print(x)   # NameError: name 'x' is not defined
```

### 4.11 Functions that use conditions and loops

````python
def is_even(n):
    return n % 2 == 0

def grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    return "F"

def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_even(10), is_even(7))
print(grade(85), grade(65), grade(20))
print(factorial(6))
print([n for n in range(1, 30) if is_prime(n)])
````

### 4.12 Functions calling other functions

````python
def square(n):
    return n * n

def sum_of_squares(a, b):
    return square(a) + square(b)

print(sum_of_squares(3, 4))
````

### 4.13 Recursion

A function that calls itself. It needs a **base case** to stop.

````python
def factorial_recursive(n):
    if n <= 1:                         # base case
        return 1
    return n * factorial_recursive(n - 1)

print(factorial_recursive(5))

def countdown(n):
    if n == 0:
        print("Liftoff!")
        return
    print(n)
    countdown(n - 1)

countdown(3)
````

### 4.14 Lambda functions

Small anonymous functions written in one line: `lambda arguments: expression`.

````python
square = lambda x: x ** 2
add = lambda a, b: a + b

print(square(5))
print(add(2, 3))

# Common use: as a key for sorting
students = [("Ali", 85), ("Sara", 92), ("Zain", 78)]
students.sort(key=lambda s: s[1], reverse=True)
print(students)
````

### 4.15 Functions as values: `map()`, `filter()`

````python
numbers = [1, 2, 3, 4, 5, 6]

doubled = list(map(lambda x: x * 2, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))

print(doubled)
print(evens)
````

### 4.16 A common beginner trap: mutable default arguments

````python
# Bad: the same list is reused on every call
def add_item_bad(item, items=[]):
    items.append(item)
    return items

print(add_item_bad("a"))
print(add_item_bad("b"))     # surprise: ['a', 'b']

# Good: use None and create a fresh list inside
def add_item_good(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item_good("a"))
print(add_item_good("b"))    # ['b']
````

### 4.17 Useful built-in functions

````python
nums = [5, 2, 9, 1]

print(len(nums))
print(sum(nums))
print(min(nums), max(nums))
print(sorted(nums))
print(sorted(nums, reverse=True))
print(abs(-7))
print(round(3.14159, 2))
print(pow(2, 5))
print(list(reversed(nums)))
print(any([False, True]), all([True, False]))
````

## 5. Putting It Together: Mini Programs

### 5.1 Number guessing game

````python
import random

secret = random.randint(1, 10)
attempts = 0

while True:
    guess = int(input("Guess a number between 1 and 10: "))
    attempts += 1

    if guess == secret:
        print(f"Correct! You took {attempts} attempt(s).")
        break
    elif guess < secret:
        print("Too low")
    else:
        print("Too high")
````

### 5.2 Simple calculator using functions

````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

def calculator():
    a = float(input("First number: "))
    op = input("Operator (+ - * /): ")
    b = float(input("Second number: "))

    if op == "+":
        print(add(a, b))
    elif op == "-":
        print(subtract(a, b))
    elif op == "*":
        print(multiply(a, b))
    elif op == "/":
        print(divide(a, b))
    else:
        print("Invalid operator")

calculator()
````

### 5.3 FizzBuzz (a classic)

````python
for i in range(1, 21):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

### 5.4 Palindrome checker

````python
def is_palindrome(text):
    cleaned = text.lower().replace(" ", "")
    return cleaned == cleaned[::-1]

for word in ["madam", "Never odd or even", "python"]:
    print(f"{word!r} -> {is_palindrome(word)}")
````

### 5.5 Student grade report

````python
def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    return "F"

students = {
    "Ali": [85, 90, 78],
    "Sara": [92, 88, 95],
    "Zain": [55, 62, 48],
}

for name, marks in students.items():
    average = sum(marks) / len(marks)
    print(f"{name:<6} average: {average:6.2f}  grade: {calculate_grade(average)}")
````
