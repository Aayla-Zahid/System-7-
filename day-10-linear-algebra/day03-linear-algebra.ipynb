## Vectors

```python
import numpy as np

a = np.array([2, 3, 4])
b = np.array([1, 0, 5])

print('a + b        :', a + b)
print('3 * a        :', 3 * a)
print('dot(a, b)    :', np.dot(a, b))
print('|a|          :', round(np.linalg.norm(a), 3))
cos = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
print('cosine sim.  :', round(cos, 3))
```

The dot product multiplies matching numbers and adds them: 2*1 + 3*0 + 4*5 = 22.

## Matrices

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print('A + B  :', A + B, sep='\n')
print('A @ B  :', A @ B, sep='\n')
print('A.T    :', A.T, sep='\n')
print('det(A) :', round(np.linalg.det(A), 3))
print('inv(A) :', np.linalg.inv(A), sep='\n')
```

- Matrix multiplication (`A @ B`) takes rows of A times columns of B. The columns of A must equal the rows of B.
- The determinant of A is -2, which is not zero, so A has an inverse.
- A dataset is a matrix: samples are rows and features are columns.

## Why it matters for ML
A linear model makes its prediction with a dot product: features dot weights, plus a bias.

```python
x = np.array([2200, 4, 3])        # area, bedrooms, bathrooms of one house
w = np.array([124, 5138, 12552])  # example weights
bias = 40000

print('prediction =', np.dot(x, w) + bias)
```
